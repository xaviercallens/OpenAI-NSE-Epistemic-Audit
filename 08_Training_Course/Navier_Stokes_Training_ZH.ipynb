## 🧮 第二模块：方程式剖析

纳维-斯托克斯方程本质上是**应用于流体的牛顿第二运动定律 ($F = ma$)**。

$$ \rho \left( \frac{\partial \mathbf{u}}{\partial t} + \mathbf{u} \cdot \nabla \mathbf{u} \right) = -\nabla p + \mu \nabla^2 \mathbf{u} + \mathbf{f} $$

1. **$\rho \frac{\partial \mathbf{u}}{\partial t}$ (非恒定加速度):** 速度随时间的变化。
2. **$\rho (\mathbf{u} \cdot \nabla \mathbf{u})$ (对流加速度):** 空间加速度，是湍流的来源，也是任何奇点可能的来源。
3. **$-\nabla p$ (压力梯度):** 流体从高压流向低压。
4. **$\mu \nabla^2 \mathbf{u}$ (粘性扩散):** 流体的“摩擦力”，使梯度变得平滑。它在三维中是否*总能*阻止速度变为无限大，正是千禧年大奖的未解问题——因此这并不是保证。
5. **$\mathbf{f}$ (外力):** 比如重力。

## 🏆 第三模块：千禧年大奖与 OpenAI 的证明

### 千禧年大奖
克雷数学研究所悬赏100万美元，寻找能够证明3D NSE的光滑解是否永远存在，或者是否会“爆破”形成**奇点**（速度变为无限大）的人。

### 正确解读 OpenAI 的结果
2026年9月，OpenAI 的多代理系统给出了一个经 Lean 4 证明助手验证的证明：在精心构造的光滑外力作用下，解*可以*在有限时间内爆破。（媒体流传的“10,000 个代理”等数字并未出现在 OpenAI 自己的论文中。）**该证明是正确的。** 它是关于不可压缩连续介质模型的定理，这正是克雷问题所问的；无外力的版本仍未解决。

### 模型在哪里不再描述真实流体
坍缩涡核的雷诺数保持在 1 左右，因此其尺寸按 $\sqrt{\nu t}$ 缩小，速度按 $\sqrt{\nu/t}$ 增长。于是模型的三个假设在同一个长度 $\ell_* = \nu/c_s$（水中约 0.7 nm，空气中约 45 nm）处**同时**失效：
- **可压缩性**：在水中，马赫数 $u/c_s$ 约在奇点前 7 皮秒超过 0.3（可压缩效应开始出现）；
- **稀薄效应**：涡核小到与分子相当（克努森数为 1 的量级）；
- **加热**：粘性加热 $\Delta T \approx u^2/c_p$ 达到数百开尔文（并非等离子体）。

在水中，**空化**出现得更早。这些都不说明证明有错：它们告诉我们，真实流体需要哪些模型中没有的物理。

### 是否有物理机制阻止它？（本项目的测量）
- 动理学理论：流体描述在分子尺度附近*终止*，但模拟表明它**不能**阻止被驱动的坍缩。
- 可压缩性与热效应：在这条路径上，它们使涡核减慢，但在 $\ell_*$ 之前不能使其停止。在另一条高雷诺数路径上，空气在局部马赫数约 0.7 处停止加速——这是对马赫数的限制，而非对速度的限制。

## 💻 第四及第五模块：粘性的作用——一维 Burgers 方程

一维 Burgers 方程 $u_t + u\,u_x = \nu\,u_{xx}$ 是 NSE 的简化“近亲”：保留了非线性对流和粘性扩散，但没有压力，且只有一维。

它清楚地展示了一个真实现象：**没有粘性 ($\nu = 0$) 时，光滑的波会变陡形成激波**——其*梯度*在破碎时刻 $t = 1/\max|u_0'|$（此处为 $t = 1$）变为无限大，而*速度*保持有界。**有粘性时**，梯度保持有限。

重要：这**不是**对 OpenAI 构造的模拟，也不能说明粘性能阻止三维纳维-斯托克斯方程爆破——一维粘性 Burgers 方程已知保持光滑，而三维情形正是未解问题，并且 OpenAI 的构造*本身包含*粘性。另外，下面的简单迎风格式本身也会带来一些数值扩散。

```python
import numpy as np
import matplotlib.pyplot as plt

# Grid setup
nx = 201
dx = 2 * np.pi / (nx - 1)
nt = 150
dt = 0.002
x = np.linspace(0, 2 * np.pi, nx)

# Initial Condition: A gentle wave
u_init = np.sin(x)

# 1. 无粘 Burgers 方程 (nu = 0)
# 对流使波变陡；梯度将在 t = 1 时变为无限大（激波）。
u_math = u_init.copy()
math_history = [u_math.copy()]

for n in range(nt):
    un = u_math.copy()
    for i in range(1, nx - 1):
        u_math[i] = un[i] - un[i] * dt / dx * (un[i] - un[i-1])
    u_math[0] = un[0] - un[0] * dt / dx * (un[0] - un[-2])
    u_math[-1] = u_math[0]
    math_history.append(u_math.copy())

# 2. 有粘 Burgers 方程 (nu > 0)
# 粘性扩散陡峭的前沿，使梯度保持有限（一维结果）。
nu = 0.1 # Physical viscosity
u_phys = u_init.copy()
phys_history = [u_phys.copy()]

for n in range(nt):
    un = u_phys.copy()
    for i in range(1, nx - 1):
        u_phys[i] = un[i] - un[i] * dt / dx * (un[i] - un[i-1]) + nu * dt / dx**2 * (un[i+1] - 2 * un[i] + un[i-1])
    u_phys[0] = un[0] - un[0] * dt / dx * (un[0] - un[-2]) + nu * dt / dx**2 * (un[1] - 2 * un[0] + un[-2])
    u_phys[-1] = u_phys[0]
    phys_history.append(u_phys.copy())

print("模拟完成！")
```

### 📊 可视化：无粘变陡 vs 有粘平滑

图中显示第 100 步，即 $t = 0.2$。红线（无粘）开始变陡；青线（有粘）明显更平滑、更低。激波本身在 $t = 1$ 时形成。

*练习：* 在上面的单元格中设置 `nt = 500` 并绘制第 500 步，观察无粘前沿趋向竖直跳跃，而有粘前沿保持平滑。

```python
plt.figure(figsize=(12, 6))
plt.style.use('dark_background')

plt.plot(x, math_history[0], 'w--', label='初始流场', alpha=0.5)

# 无粘 Burgers：趋向激波（梯度爆破）
plt.plot(x, math_history[100], 'r-', lw=2, label='无粘 (nu = 0)：变陡')

# 有粘 Burgers：梯度保持有限
plt.plot(x, phys_history[100], 'c-', lw=3, label='有粘 (nu = 0.1)：平滑')

plt.title('一维 Burgers 方程 t = 0.2：无粘变陡 vs 有粘平滑', fontsize=14)
plt.xlabel('空间 (x)')
plt.ylabel('速度 (u)')
plt.legend(facecolor='black', fontsize=12)
plt.grid(color='#333333')
plt.show()
```
