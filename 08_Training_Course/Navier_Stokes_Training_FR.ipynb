## 🏛️ Module 1: Histoire et Importance

### Qui l'a découvert ?
Les équations de Navier-Stokes ont été développées au début du 19ème siècle.
- **Claude-Louis Navier (1822):** Un ingénieur français qui a introduit le concept de viscosité.
- **George Gabriel Stokes (1845):** Un mathématicien irlandais qui a perfectionné les travaux de Navier.

### Pourquoi est-ce important ?
Elles régissent le mouvement de presque tous les fluides de l'univers.
- **L'Aéronautique** pour concevoir les ailes d'avion.
- **La Météorologie** pour prédire les ouragans.
- **Les Équipes de Formule 1** pour l'aérodynamisme.
- **La Biomédecine** pour modéliser le flux sanguin.

## 🧮 Module 2: Anatomie de l'Équation

L'équation de Navier-Stokes est essentiellement la **Deuxième Loi de Newton ($F = ma$) appliquée aux fluides**.

$$ \rho \left( \frac{\partial \mathbf{u}}{\partial t} + \mathbf{u} \cdot \nabla \mathbf{u} \right) = -\nabla p + \mu \nabla^2 \mathbf{u} + \mathbf{f} $$

Décomposons-la :
1. **$\rho \frac{\partial \mathbf{u}}{\partial t}$ (Accélération instationnaire):** Variation de la vitesse au cours du temps.
2. **$\rho (\mathbf{u} \cdot \nabla \mathbf{u})$ (Accélération convective):** Accélération spatiale du fluide (moteur de la turbulence, et source de toute singularité éventuelle).
3. **$-\nabla p$ (Gradient de pression):** Les fluides vont des hautes vers les basses pressions.
4. **$\mu \nabla^2 \mathbf{u}$ (Diffusion visqueuse):** La "friction", qui lisse les gradients. Savoir si elle suffit *toujours* à empêcher des vitesses infinies en 3D est justement la question ouverte du Prix du Millénaire : ce n'est donc pas une garantie.
5. **$\mathbf{f}$ (Forces externes):** Gravité, etc.

## 🏆 Module 3: Le Prix du Millénaire & la Preuve d'OpenAI

### Le Prix du Millénaire
Le Clay Mathematics Institute offre 1 000 000 $ pour prouver si des solutions lisses des ENS en 3D existent toujours, ou si elles peuvent "exploser" en une **singularité** (vitesse infinie).

### Le résultat d'OpenAI, bien lu
En septembre 2026, un système multi-agents d'OpenAI a produit une preuve, vérifiée par l'assistant de preuve Lean 4, qu'avec une force extérieure lisse soigneusement construite, une solution *peut* exploser en temps fini. (Les chiffres de presse comme « 10 000 agents » n'apparaissent pas dans les publications d'OpenAI.) **La preuve est correcte.** C'est un théorème sur le modèle continu incompressible — celui du problème du Clay ; la version sans force reste ouverte.

### Où le modèle cesse de décrire un fluide réel
Le cœur du tourbillon garde un nombre de Reynolds proche de 1 : sa taille décroît comme $\sqrt{\nu t}$ et sa vitesse croît comme $\sqrt{\nu/t}$. Trois hypothèses du modèle tombent alors **ensemble**, à une seule longueur $\ell_* = \nu/c_s$ (environ 0,7 nm dans l'eau, 45 nm dans l'air) :
- **Compressibilité** : le nombre de Mach $u/c_s$ dépasse 0,3 (début des effets de compressibilité) environ 7 picosecondes avant la singularité dans l'eau ;
- **Raréfaction** : le cœur devient aussi petit que les molécules (nombre de Knudsen d'ordre un) ;
- **Échauffement** : l'échauffement visqueux $\Delta T \approx u^2/c_p$ atteint quelques centaines de kelvins (pas un plasma).

Dans l'eau, la **cavitation** arrive encore plus tôt. Rien de cela ne rend la preuve fausse : cela indique quelle physique, absente du modèle, un fluide réel ferait intervenir.

### Quelque chose de physique l'arrête-t-il ? (ce que le projet a mesuré)
- Théorie cinétique : la description fluide *se termine* près de l'échelle moléculaire, mais une simulation montre qu'elle **n'arrête pas** un effondrement entretenu.
- Compressibilité et chaleur : sur cette route, elles ralentissent le cœur sans l'arrêter avant $\ell_*$. Sur une autre route, à grand nombre de Reynolds, l'air cesse d'accélérer vers un Mach local d'environ 0,7 — une limite sur le nombre de Mach, pas sur la vitesse.

## 💻 Module 4 & 5: La viscosité en action — l'équation de Burgers 1D

L'équation de Burgers 1D $u_t + u\,u_x = \nu\,u_{xx}$ est une cousine simplifiée des ENS : elle garde la convection non linéaire et la diffusion visqueuse, sans pression et en une seule dimension.

Elle montre clairement un phénomène réel : **sans viscosité ($\nu = 0$), une onde lisse se raidit en choc** — son *gradient* devient infini à l'instant de déferlement $t = 1/\max|u_0'|$ (ici $t = 1$), tandis que sa *vitesse* reste bornée. **Avec viscosité**, le gradient reste fini.

Important : ce n'est **pas** une simulation de la construction d'OpenAI, et cela ne montre pas que la viscosité empêche l'explosion de Navier-Stokes en 3D — en 1D, Burgers visqueux reste régulier, alors qu'en 3D c'est la question ouverte, et la construction d'OpenAI *inclut* la viscosité. Le schéma décentré ci-dessous ajoute aussi un peu de diffusion numérique.

```python
import numpy as np
import matplotlib.pyplot as plt

# Grid setup
nx = 201
dx = 2 * np.pi / (nx - 1)
nt = 150
dt = 0.002
x = np.linspace(0, 2 * np.pi, nx)

# Initial Condition: A gentle wave
u_init = np.sin(x)

# 1. Burgers non visqueux (nu = 0)
# La convection raidit l'onde ; le gradient deviendrait infini à t = 1 (choc).
u_math = u_init.copy()
math_history = [u_math.copy()]

for n in range(nt):
    un = u_math.copy()
    for i in range(1, nx - 1):
        u_math[i] = un[i] - un[i] * dt / dx * (un[i] - un[i-1])
    u_math[0] = un[0] - un[0] * dt / dx * (un[0] - un[-2])
    u_math[-1] = u_math[0]
    math_history.append(u_math.copy())

# 2. Burgers visqueux (nu > 0)
# La viscosité diffuse le front et garde le gradient fini (résultat 1D).
nu = 0.1 # Viscosité physique
u_phys = u_init.copy()
phys_history = [u_phys.copy()]

for n in range(nt):
    un = u_phys.copy()
    for i in range(1, nx - 1):
        u_phys[i] = un[i] - un[i] * dt / dx * (un[i] - un[i-1]) + nu * dt / dx**2 * (un[i+1] - 2 * un[i] + un[i-1])
    u_phys[0] = un[0] - un[0] * dt / dx * (un[0] - un[-2]) + nu * dt / dx**2 * (un[1] - 2 * un[0] + un[-2])
    u_phys[-1] = u_phys[0]
    phys_history.append(u_phys.copy())

print("Simulation terminée!")
```

### 📊 Visualisation : raidissement non visqueux vs lissage visqueux

Le graphique montre l'étape 100, soit $t = 0{,}2$. La ligne rouge (non visqueuse) commence à se raidir ; la ligne cyan (visqueuse) est plus lisse et plus basse. Le choc lui-même se forme à $t = 1$.

*Exercice :* mettez `nt = 500` dans la cellule précédente et tracez l'étape 500 pour voir le front non visqueux tendre vers une marche verticale, tandis que le front visqueux reste lisse.

```python
plt.figure(figsize=(12, 6))
plt.style.use('dark_background')

plt.plot(x, math_history[0], 'w--', label='Flux Initial', alpha=0.5)

# Burgers non visqueux : raidissement vers un choc
plt.plot(x, math_history[100], 'r-', lw=2, label='Non visqueux (nu = 0) : raidissement')

# Burgers visqueux : gradient fini
plt.plot(x, phys_history[100], 'c-', lw=3, label='Visqueux (nu = 0,1) : lissé')

plt.title('Burgers 1D à t = 0,2 : raidissement non visqueux vs lissage visqueux', fontsize=14)
plt.xlabel('Espace (x)')
plt.ylabel('Vitesse (u)')
plt.legend(facecolor='black', fontsize=12)
plt.grid(color='#333333')
plt.show()
```
