## 🧮 Module 2: Anatomie de l'Équation

L'équation de Navier-Stokes est essentiellement la **Deuxième Loi de Newton ($F = ma$) appliquée aux fluides**.

$$ \rho \left( \frac{\partial \mathbf{u}}{\partial t} + \mathbf{u} \cdot \nabla \mathbf{u} \right) = -\nabla p + \mu \nabla^2 \mathbf{u} + \mathbf{f} $$

Décomposons-la :
1. **$\rho \frac{\partial \mathbf{u}}{\partial t}$ (Accélération instationnaire):** Variation de la vitesse au cours du temps.
2. **$\rho (\mathbf{u} \cdot \nabla \mathbf{u})$ (Accélération convective):** Accélération spatiale du fluide (cause la turbulence).
3. **$-\nabla p$ (Gradient de pression):** Les fluides vont des hautes vers les basses pressions.
4. **$\mu \nabla^2 \mathbf{u}$ (Diffusion visqueuse):** La "friction". Empêche les vitesses infinies (Censure Thermodynamique).
5. **$\mathbf{f}$ (Forces externes):** Gravité, etc.

## 💻 Module 4 & 5: Résolution avec la Régularisation Multi-Échelle (LeanFlow)

Pour empêcher l'IA de générer des singularités non physiques, nous utilisons la **Régularisation Multi-Échelle (LeanFlow)**. 

Écrivons une simulation Python de l'équation de Burgers 1D (un cousin des ENS) pour démontrer la théorie d'OpenAI face à la Réalité.

```python
import numpy as np
import matplotlib.pyplot as plt

# Grid setup
nx = 201
dx = 2 * np.pi / (nx - 1)
nt = 150
dt = 0.002
x = np.linspace(0, 2 * np.pi, nx)

# Initial Condition: A gentle wave
u_init = np.sin(x)

# 1. Mathématiques Pures - OpenAI Singularity Claim
u_math = u_init.copy()
math_history = [u_math.copy()]

for n in range(nt):
    un = u_math.copy()
    for i in range(1, nx - 1):
        u_math[i] = un[i] - un[i] * dt / dx * (un[i] - un[i-1])
    u_math[0] = un[0] - un[0] * dt / dx * (un[0] - un[-2])
    u_math[-1] = u_math[0]
    math_history.append(u_math.copy())

# 2. Solveur Multi-Échelle LeanFlow (Censure Thermodynamique)
nu = 0.1 # Viscosité physique
u_phys = u_init.copy()
phys_history = [u_phys.copy()]

for n in range(nt):
    un = u_phys.copy()
    for i in range(1, nx - 1):
        u_phys[i] = un[i] - un[i] * dt / dx * (un[i] - un[i-1]) + nu * dt / dx**2 * (un[i+1] - 2 * un[i] + un[i-1])
    u_phys[0] = un[0] - un[0] * dt / dx * (un[0] - un[-2]) + nu * dt / dx**2 * (un[1] - 2 * un[0] + un[-2])
    u_phys[-1] = u_phys[0]
    phys_history.append(u_phys.copy())

print("Simulation terminée!")
```

### 📊 Visualisation : Explosion Mathématique vs Réalité Physique

Observez comment la ligne rouge (modèle OpenAI) crée une singularité aux pentes infinies. La ligne bleue (LeanFlow) lisse la courbe, prouvant que la viscosité sauve l'univers d'une explosion mathématique.

```python
plt.figure(figsize=(12, 6))
plt.style.use('dark_background')

plt.plot(x, math_history[0], 'w--', label='Flux Initial', alpha=0.5)

# OpenAI Singularity
plt.plot(x, math_history[100], 'r-', lw=2, label='Singularité Mathématique')

# LeanFlow
plt.plot(x, phys_history[100], 'c-', lw=3, label='Réalité LeanFlow (Bornée)')

plt.title('Formation: Singularité Abstraite vs Régularisation Multi-Échelle', fontsize=14)
plt.xlabel('Espace (x)')
plt.ylabel('Vitesse (u)')
plt.legend(facecolor='black', fontsize=12)
plt.grid(color='#333333')
plt.show()
```
