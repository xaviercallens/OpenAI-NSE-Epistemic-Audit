### 💻 Interactive Experiment / Expérience Interactive / 交互式实验

🇺🇸 Below is a small solver simulating a 1D viscous fluid (Burgers' equation). You can see how a mathematical singularity tries to form (the wave steepens), but physical viscosity (Thermodynamic Censorship) diffuses it, preventing an infinite blow-up.

🇫🇷 Ci-dessous se trouve un petit solveur simulant un fluide visqueux en 1D (équation de Burgers). Vous pouvez voir comment une singularité mathématique tente de se former (la vague se raidit), mais la viscosité physique (Censure Thermodynamique) la diffuse, empêchant une explosion infinie.

🇨🇳 下面是一个模拟一维粘性流体（伯格斯方程）的小型求解器。您可以看到数学奇点试图形成（波变陡），但物理粘性（热力学审查）将其扩散，从而防止了无限的能量爆发。

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1D Viscous Burgers' Equation Solver (Small LeanFlow Concept) ---
# du/dt + u * du/dx = nu * d^2u/dx^2

nx = 101       # Number of spatial grid points
dx = 2 * np.pi / (nx - 1)
nt = 150       # Number of time steps
dt = 0.005     # Time step size
nu = 0.05      # Viscosity (Physical limit / Thermodynamic Censorship)

x = np.linspace(0, 2 * np.pi, nx)
u = np.sin(x)  # Initial condition: Smooth wave
u_history = [u.copy()]

for n in range(nt):
    un = u.copy()
    for i in range(1, nx - 1):
        # Convection term (creates singularity) + Diffusion term (Censorship)
        u[i] = un[i] - un[i] * dt / dx * (un[i] - un[i-1]) + nu * dt / dx**2 * (un[i+1] - 2 * un[i] + un[i-1])
    
    # Periodic boundary conditions
    u[0] = un[0] - un[0] * dt / dx * (un[0] - un[-2]) + nu * dt / dx**2 * (un[1] - 2 * un[0] + un[-2])
    u[-1] = u[0]
    u_history.append(u.copy())

print("Simulation completed! / Simulation terminée! / 模拟完成！")
```

```python
# Plotting the results over time
plt.figure(figsize=(10, 6))
plt.style.use('dark_background')
plt.plot(x, u_history[0], 'w--', label='t=0 (Initial Smooth Wave)')
plt.plot(x, u_history[50], 'y-', label='t=50 (Shock forming / Singularité en approche)')
plt.plot(x, u_history[-1], 'c-', linewidth=3, label='t=150 (Viscosity dissipates energy / Censure Thermodynamique)')

plt.title('1D Viscous Fluid Solver: Viscosity Prevents Mathematical Singularity', fontsize=14)
plt.xlabel('Space (x)', fontsize=12)
plt.ylabel('Velocity (u)', fontsize=12)
plt.legend(facecolor='black')
plt.grid(color='#333333')
plt.show()
```
