### 📊 Step 4: Interactive Visual Telemetry & Decision Boundary / Visualisation Interactive / 交互式可视化遥测

🇺🇸 Plot the decision boundary showing why OpenAI's unconstrained Sobolev trajectory is REJECTED while LeanFlow Dual-Scale trajectory is PASSED.

🇫🇷 Tracez les frontières de décision montrant pourquoi la trajectoire de Sobolev d'OpenAI est REJETÉE alors que LeanFlow est ACCEPTÉ.

🇨🇳 绘制决策边界图，展示为什么 OpenAI 的无约束 Sobolev 轨迹被拒绝，而 LeanFlow 双尺度轨迹通过验证。


```python
# --- Step 4: Visual Decision Boundary Plotting ---
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1.0, 300)
v_openai = 10.0 + 440.0 * (t ** 4)         # Accelerates past 343 m/s (Mach 1.31)
v_leanflow = 10.0 + 75.0 * (1 - np.exp(-3*t)) # Saturation at 85 m/s (Mach 0.25)

mach_openai = v_openai / 343.0
mach_leanflow = v_leanflow / 343.0

plt.figure(figsize=(12, 6))
plt.style.use('dark_background')

plt.plot(t, mach_openai, 'r--', linewidth=2.5, label='OpenAI Proof Search Trajectory (Breaches Ma > 0.3 Limit)')
plt.plot(t, mach_leanflow, 'g-', linewidth=3.0, label='LeanFlow Dual-Scale Trajectory (Physical Boundary)')
plt.axhline(y=0.3, color='yellow', linestyle='--', linewidth=2, label='Incompressible Model Limit (Mach 0.3)')

plt.fill_between(t, 0, 0.3, color='green', alpha=0.15, label='Physically Admissible Zone')
plt.fill_between(t, 0.3, 1.5, color='red', alpha=0.15, label='Unphysical Model Breakdown Zone')

plt.title('OpenAI PI-FPS PoC: Physical Admissibility Decision Boundary', fontsize=14, fontweight='bold')
plt.xlabel('Time t (normalized)', fontsize=12)
plt.ylabel('Local Mach Number (Ma)', fontsize=12)
plt.legend(facecolor='black', edgecolor='grey')
plt.grid(alpha=0.3)
plt.show()
```
