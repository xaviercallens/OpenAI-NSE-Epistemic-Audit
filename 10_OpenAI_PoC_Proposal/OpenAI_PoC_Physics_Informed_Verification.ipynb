### 📊 Step 4: Schematic Mach Trajectories / Trajectoires schématiques / 示意马赫数轨迹

🇺🇸 A **schematic** plot of two made-up Mach-number histories against the Mach 0.3 line (onset of compressibility effects). It illustrates what a validity label reports; the curves are not OpenAI's construction, not a proof-search trajectory and not a LeanFlow run.

🇫🇷 Un tracé **schématique** de deux historiques de Mach inventés, comparés à la ligne Mach 0,3 (début des effets de compressibilité). Il illustre ce que rapporte une étiquette de validité ; les courbes ne sont ni la construction d'OpenAI, ni une trajectoire de recherche de preuve, ni un calcul LeanFlow.

🇨🇳 两条人为构造的马赫数历史曲线与马赫数 0.3 线（可压缩效应开始出现）的**示意图**。它说明有效性标注所报告的内容；这些曲线既不是 OpenAI 的构造，也不是证明搜索轨迹，更不是 LeanFlow 计算。

```python
# --- Step 4: SCHEMATIC Mach histories (made-up shapes) ---
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1.0, 300)
v_openai = 10.0 + 440.0 * (t ** 4)         # made-up accelerating history, ends at Mach 1.31
v_leanflow = 10.0 + 75.0 * (1 - np.exp(-3*t)) # made-up saturating history, ends near Mach 0.25

mach_openai = v_openai / 343.0
mach_leanflow = v_leanflow / 343.0

plt.figure(figsize=(12, 6))
plt.style.use('dark_background')

plt.plot(t, mach_openai, 'r--', linewidth=2.5, label='Illustrative accelerating state (leaves the model range)')
plt.plot(t, mach_leanflow, 'g-', linewidth=3.0, label='Illustrative subsonic state (inside the model range)')
plt.axhline(y=0.3, color='yellow', linestyle='--', linewidth=2, label='Onset of compressibility effects (Mach 0.3)')

plt.fill_between(t, 0, 0.3, color='green', alpha=0.15, label='Incompressible model applies')
plt.fill_between(t, 0.3, 1.5, color='red', alpha=0.15, label='Incompressible model no longer applies')

plt.title('Model-validity labelling (schematic)', fontsize=14, fontweight='bold')
plt.xlabel('Time t (normalized)', fontsize=12)
plt.ylabel('Local Mach Number (Ma)', fontsize=12)
plt.legend(facecolor='black', edgecolor='grey')
plt.grid(alpha=0.3)
plt.show()
```
